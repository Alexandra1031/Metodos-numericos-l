El método de Newton-Raphson es un método numérico iterativo que permite aproximar las raíces de una función utilizando su derivada. Consiste en partir de una aproximación inicial y calcular nuevas aproximaciones mediante la fórmula:

$$p_n = p_{n-1} -
\frac{f(p_{n-1})}{f'(p_{n-1})},
\qquad n \geq 1
$$


En esta tarea se implementa el algoritmo 2.3 del libro de Burden y se comprueba su funcionamiento mediante el ejemplo 1 de la página 51, que consiste en encontrar una raíz de la ecuación $\cos x-x=0$. Se utiliza como aproximación inicial $p_0=\pi/4$ y se realizan cuatro iteraciones.


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import sympy as sp
```

**numpy:** permite realizar operaciones numéricas.

**matplotlib.pyplot:** permite graficar funciones.

**pandas:** permite organizar los resultados en tablas.

**sympy:** permite trabajar con expresiones simbólicas.

```python
# Definimos la funcion
def f(x):
   return np.cos(x) - x
```


La función que se desea resolver es:

$$
f(x)=\cos x-x
$$

Encontrar una raíz consiste en determinar un valor $p$ tal que $f(p)=0$.


###**Grafica de la funcion**

```python
xx = np.linspace(0,1.2, 100) #genera 100 puntos entre 1.3 y 2
plt.plot(xx, f(xx)) #grafica la función
plt.axhline(y=0, color="pink") #dibuja el eje horizontal y=0,para visualizar dónde se encuentra la raíz.
plt.show()
```


Se grafica la función en el intervalo $[0,1.2]$. La curva azul representa $f(x)=\cos x-x$ y la recta rosa representa $y=0$. La intersección de ambas permite visualizar la raíz que se aproximará mediante el método de Newton



```python
# Definimos la derivada
def df(x):
  return -np.sin(x) - 1
```


Derivando la función original, se obtiene:

$$
f'(x)=-\sin x-1
$$

La derivada se utiliza para calcular las nuevas aproximaciones mediante la fórmula de Newton


```python

# Metodo de Newton-Raphson
def newton(p0, n):
    lista = []
    error = []

    for i in range(n):
        fp0 = f(p0)
        dif = df(p0)

        p = p0 - (fp0/dif)
        lista.append(p)

        e = np.abs(p - p0)
        error.append(e)

        p0 = p

    return lista, error


# Ejecutamos el metodo
p = newton(np.pi/4, 4)

# Mostramos la tabla de resultados

# Mostramos la tabla de resultados
tabla = pd.DataFrame({
    "p_n": p[0],
    "Error absoluto": p[1]
})

tabla.index.name = "Iteracion"
tabla

```


La función newton(p0, n) recibe la aproximación inicial $p_0$ y el número de iteraciones $n$. En cada iteración calcula una nueva aproximación mediante:

$$
p_n=p_{n-1}-\frac{f(p_{n-1})}{f'(p_{n-1})}
$$

Posteriormente, calcula el error absoluto:

$$
E_n=|p_n-p_{n-1}|
$$

Las aproximaciones anteriores, las nuevas aproximaciones y los errores se almacenan en listas para analizar la convergencia del método.



Se toma como aproximación inicial:

$$
p_0=\frac{\pi}{4}\approx0.7853981634
$$

Se realizan cuatro iteraciones, como en el ejemplo 1 del libro de Burden, para obtener una aproximación de la raíz.



La tabla permite observar la evolución de las aproximaciones y el error absoluto en cada iteración. A medida que avanza el procedimiento, las aproximaciones se acercan a la raíz y el error disminuye, lo que nos muetra la convergencia del método.



## Conclusión

Mediante el método de Newton-Raphson se aproximó la raíz de la ecuación $\cos x-x=0$. Después de cuatro iteraciones, se obtiene una aproximación de:

$$\boxed{p\approx0.7390851332}$$
