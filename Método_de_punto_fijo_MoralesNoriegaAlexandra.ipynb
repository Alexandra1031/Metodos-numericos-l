El **método de iteración de punto fijo** es un método numérico que permite aproximar soluciones de ecuaciones mediante su transformación a la forma x=g(x). A partir de una aproximación inicial, se genera una sucesión de valores utilizando la fórmula:

$$
p_n=g(p_{n-1}).
$$

En esta actividad se implementará en Python el Algoritmo 2.2 de Burden y Faires para resolver la ecuación:

$$
x^3+4x^2-10=0.
$$

Para ello, se utilizará la función $(g(x)=\frac{1}{2}\sqrt{10-x^3})$. Antes de aplicar el algoritmo, se verificarán las condiciones del Teorema del Punto Fijo para justificar la convergencia del método. Y al final, se analizarán las aproximaciones obtenidas y se comprobará la solución en la ecuación original.

## Verificación del Teorema del Punto Fijo

Para el inciso (c) se tiene:

$$
g(x)=\frac{1}{2}\sqrt{10-x^3}.
$$

Tomamos el intervalo:

$$
[a,b]=[1,1.5].
$$

Primero verificamos que:

$$
g([1,1.5])\subseteq[1,1.5].
$$

Evaluamos en los extremos:

$$
g(1)=1.5
$$

y

$$
g(1.5)\approx1.286954.
$$

Como \(g(x)\) es decreciente en el intervalo,

$$
1.286954\leq g(x)\leq1.5.
$$

Por lo tanto:

$$
g([1,1.5])\subseteq[1,1.5].
$$

Además, existe \(g'(x)\) en \((1,1.5)\), donde

$$
g'(x)=-\frac{3x^2}{4\sqrt{10-x^3}}.
$$

Así,

$$
|g'(x)|=\frac{3x^2}{4\sqrt{10-x^3}}.
$$

El máximo en el intervalo se obtiene en \(x=1.5\):

$$
|g'(1.5)|\approx0.655618.
$$

Por lo tanto, podemos tomar:

$$
k=0.655618<1,
$$

de modo que:

$$
|g'(x)|\leq k<1,
\qquad \forall x\in(1,1.5).
$$
y

$$
p_0=1.5\in[1,1.5].
$$

Entonces, se cumplen las condiciones del Teorema del Punto Fijo.
Por lo tanto, la sucesión

$$
p_n=g(p_{n-1}),\qquad n\geq1
$$

converge al único punto fijo

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Funcion original
def funcion1(x):
    return x**3 + 4*x**2 - 10


# Funcion de punto fijo
def funcion2(x):
    return 0.5 * math.sqrt(10 - x**3)
```



La función funcion1(x) representa la ecuación original:

$$
f(x)=x^3+4x^2-10.
$$

Y la función funcion2(x) representa la función de punto fijo:

$$
g(x)=\frac{1}{2}\sqrt{10-x^3}.
$$

En Python, el operador ** representa una potencia y math.sqrt() calcula la raíz cuadrada. Para utilizar esta última instrucción, se importa la biblioteca mediante import math.

Finalmente, se utilizan las bibliotecas numpy y matplotlib.pyplot para generar los valores y graficar la función original, permitiendo visualizar su comportamiento y aproximar su raíz.

```python
# Ingreso datos de entrada para el metodo
p0 = 1.5

# Guardar valor inicial
p00 = p0

# Guardar valores iniciales del error y del numero de iteraciones
tol = 0.00000001
max = 100
error = 100
niter = 1


# Metodo de punto fijo

print("\n\t n \t\t p0 \t\t Pn \t\t f(Pn) \t\t error")

# Ciclo iterativo
while error > tol and niter <= max:

    # Evaluo siguiente valor
    p = funcion2(p0)

    # Evaluacion de la funcion en el punto p
    fp = funcion1(p)

    # Calculo del error
    error = abs(p - p0)

    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f}".format(
        niter, p0, p, fp, error
    ))

    # Actualizar valor de p0
    p0 = p

    # Incrementar numero de iteracion
    niter += 1


print("\nLa raiz de la funcion dada es {0:10.8f}".format(p))
```

```python
x = np.linspace(0, 3)  # Genera una particion
plt.plot(x, funcion1(x))  # Grafica las parejas ordenadas (x, f(x))
plt.title('Grafica de f(x) = x^3 + 4x^2 - 10')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(y=0, color='pink')  # Linea horizontal en y = 0
plt.axvline(x=0, color='purple')  # Linea vertical en x = 0
plt.grid(True)
plt.show()
```

La intersección de la curva azul con el eje horizontal permite identificar gráficamente la raíz de la ecuación, cuya aproximación es $p\approx1.36523001$.
