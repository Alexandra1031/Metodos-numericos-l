# **ERROR** **DE** **MAQUINA**

Este código sirve para calcular el épsilon de la máquina,  que representa el número más pequeño que la computadora puede sumar a 1.0 de forma que el resultado siga siendo mayor que 1.0.

```python
def epsilon():
    epsilons = []
    E = 1.0
    while 1.0 + (E / 2.0) > 1.0:
        epsilons.append(E)
        E = E / 2
    epsilons.append(E)
    cero_mas_E = 0.0 + E
    return epsilons, cero_mas_E


if __name__ == "__main__":
    lista_epsilons, resultado_final = epsilon()
    print(lista_epsilons)
    print(resultado_final)
```
